Short walkthrough of scalars, vectors, covectors, bilinear forms, and change of basis.
Requires an editable install: `pip install -e ".[dev,notebook]"`.


```python
from tensor_ladder import (
    Basis, Vector, Covector, BilinearForm, ChangeOfBasis, Scalar, __version__
)
import numpy as np
print("tensor_ladder", __version__)
```

## Pairing $\omega(v) = \omega_i v^i$


```python
e = Basis.standard(2, name="e")
v = Vector.from_iterable([3.0, 4.0], e)
omega = Covector.from_iterable([1.0, 2.0], e)
print(omega(v))
assert float(omega(v)) == 11.0
```

## Euclidean metric and flat


```python
g = BilinearForm.euclidean(2, e)
print("g(v,v) =", g(v, v))
print("v flat =", g.flat(v))
```

## Change of basis — pairing invariant


```python
e_new = Basis(np.array([[2.0, 0.0], [0.0, 1.0]]), name="e-tilde")
change = ChangeOfBasis.from_bases(e, e_new)
v_new, omega_new = v.in_basis(change), omega.in_basis(change)
print(v_new)
print(omega_new)
print("old", float(omega(v)), "new", float(omega_new(v_new)))
assert np.isclose(float(omega(v)), float(omega_new(v_new)))
```
