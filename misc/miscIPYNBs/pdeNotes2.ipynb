# Explicit Step-by-Step Solutions  
## Transport Equation, Transport with Decay, Heat Equation, and Wave Equation

These are the basic cases: pure initial-value problems (infinite line) and initial-boundary-value problems (finite interval). Every major algebraic step is shown.

---



## 1. Transport Equation

### Equation
$$
u_t + c\, u_x = 0, \qquad -\infty < x < \infty,\quad t > 0,
$$
with initial condition
$$
u(x,0) = \phi(x).
$$



### Solution by characteristics

**Step 1.** The characteristic curves satisfy
$$
\frac{dx}{dt} = c \quad\Rightarrow\quad x - c t = \text{constant}.
$$
Along each of these lines the solution is constant.

**Step 2.** Therefore the general solution is an arbitrary function of the characteristic variable:
$$
u(x,t) = f(x - c t).
$$

**Step 3.** Apply the initial condition:
$$
u(x,0) = f(x) = \phi(x).
$$
Hence
$$
u(x,t) = \phi(x - c t).
$$

**Interpretation.** The initial profile is simply translated to the right (if $c>0$) at speed $c$ without changing shape.

---



## 2. Transport Equation with Decay


### Equation
$$
u_t + c\, u_x + a u = 0, \qquad -\infty < x < \infty,\quad t > 0,
$$
with
$$
u(x,0) = \phi(x).
$$
($a>0$ is the decay rate.)

### Solution

**Step 1.** Characteristics are still the lines $x - c t = \xi$ (constant).

**Step 2.** Along a characteristic, the PDE reduces to an ODE. Parametrize by $t$ with $x = \xi + c t$. Then
$$
\frac{d}{dt}u(\xi + c t,\, t) + a\, u = 0.
$$
This is
$$
\frac{dv}{dt} + a v = 0 \quad\Rightarrow\quad v(t) = v(0)\, e^{-a t}.
$$

**Step 3.** At $t=0$ we have $v(0)=\phi(\xi)$, so
$$
u(\xi + c t,\, t) = \phi(\xi)\, e^{-a t}.
$$
Replace $\xi = x - c t$:
$$
u(x,t) = \phi(x - c t)\, e^{-a t}.
$$

**Interpretation.** The profile is translated at speed $c$ and simultaneously multiplied by the decaying exponential $e^{-a t}$.

---



## 3. Heat Equation



### 3.1 Pure initial-value problem (infinite line)

$$
u_t = k u_{xx}, \qquad -\infty < x < \infty,\quad t > 0,
$$
$$
u(x,0) = \phi(x).
$$

**Solution (fundamental solution / convolution).**  
The fundamental solution (heat kernel) is
$$
G(x,t) = \frac{1}{\sqrt{4\pi k t}}\, \exp\Bigl(-\frac{x^2}{4 k t}\Bigr).
$$
The solution is the convolution
$$
u(x,t) = \int_{\chi \ = \ -\infty}^{\infty} G(x-\chi,t)\, \phi(\chi)\, d\chi
= \frac{1}{\sqrt{4\pi k t}} \int_{-\infty}^{\infty} \exp\Bigl(-\frac{(x-\chi)^2}{4 k t}\Bigr) \phi(\chi)\, d\chi.
$$

**Steps that produce the kernel (sketch).**  
Fourier transform in $x$: the PDE becomes an ODE in $t$ for each frequency $\xi$. Solving and inverting the transform yields the Gaussian above.



### 3.2 Finite interval with Dirichlet boundary conditions

$$
u_t = k u_{xx}, \qquad 0 < x < L,\quad t > 0,
$$
$$
u(0,t)=u(L,t)=0, \qquad u(x,0)=\phi(x).
$$

**Step 1 – Separation.** Assume $u=X(x)T(t)$. Then
$$
\frac{T'}{k T} = \frac{X''}{X} = -\lambda.
$$

**Step 2 – Eigenvalue problem.**  
$X(0)=X(L)=0$ forces
$$
\lambda_n = \Bigl(\frac{n\pi}{L}\Bigr)^2, \qquad X_n(x)=\sin\Bigl(\frac{n\pi x}{L}\Bigr),\quad n=1,2,\dots
$$

**Step 3 – Time factors.**
$$
T_n(t) = A_n\, e^{-k\lambda_n t}.
$$

**Step 4 – Superposition.**
$$
u(x,t) = \sum_{n=1}^\infty A_n \sin\Bigl(\frac{n\pi x}{L}\Bigr) e^{-k(n\pi/L)^2 t}.
$$

**Step 5 – Coefficients.**  
The initial condition is the Fourier sine series of $\phi$:
$$
A_n = \frac{2}{L}\int_0^L \phi(x)\sin\Bigl(\frac{n\pi x}{L}\Bigr)\, dx.
$$

## 4. Wave Equation


### 4.1 Pure initial-value problem (infinite line) – d’Alembert’s formula

$$
u_{tt} = c^2 u_{xx}, \qquad -\infty < x < \infty,\quad t > 0,
$$
$$
u(x,0)=\phi(x), \qquad u_t(x,0)=\psi(x).
$$

**Step 1 – Factor the operator.**  
$$
(\partial_t - c\partial_x)(\partial_t + c\partial_x)u = 0.
$$
Introduce the characteristic coordinates
$$
\xi = x - c t, \qquad \eta = x + c t.
$$

**Step 2 – General solution.**  
In these coordinates the equation becomes $u_{\xi\eta}=0$, whose general solution is
$$
u(x,t) = f(x-ct) + g(x+ct).
$$

**Step 3 – Apply initial conditions.**  
$$
f(x)+g(x)=\phi(x),
$$
$$
-c f'(x)+c g'(x)=\psi(x).
$$
Integrate the second equation and solve for $f$ and $g$. The result is d’Alembert’s formula:
$$
u(x,t) = \frac12\bigl(\phi(x-ct)+\phi(x+ct)\bigr) + \frac{1}{2c}\int_{x-ct}^{x+ct}\psi(s)\,ds.
$$

**Interpretation.** The solution is the average of the initial displacement traveling left and right, plus the accumulated effect of the initial velocity.



### 4.2 Finite string with fixed ends (Dirichlet)

$$
u_{tt} = c^2 u_{xx}, \qquad 0 < x < L,\quad t > 0,
$$
$$
u(0,t)=u(L,t)=0,
$$
$$
u(x,0)=\phi(x), \qquad u_t(x,0)=\psi(x).
$$

**Step 1 – Separation.** $u=X(x)T(t)$ yields
$$
\frac{T''}{c^2 T} = \frac{X''}{X} = -\lambda.
$$

**Step 2 – Eigenvalue problem.**  
$X(0)=X(L)=0$ gives
$$
\lambda_n = \Bigl(\frac{n\pi}{L}\Bigr)^2, \qquad X_n=\sin\Bigl(\frac{n\pi x}{L}\Bigr),\quad n=1,2,\dots
$$

**Step 3 – Time factors.**
$$
T_n(t) = A_n\cos(\omega_n t) + B_n\sin(\omega_n t), \qquad \omega_n = \frac{n\pi c}{L}.
$$

**Step 4 – Superposition.**
$$
u(x,t) = \sum_{n=1}^\infty \sin\Bigl(\frac{n\pi x}{L}\Bigr)\bigl(A_n\cos(\omega_n t)+B_n\sin(\omega_n t)\bigr).
$$

**Step 5 – Coefficients.**
$$
A_n = \frac{2}{L}\int_0^L\phi(x)\sin\Bigl(\frac{n\pi x}{L}\Bigr)\,dx,
$$
$$
B_n = \frac{2}{n\pi c}\int_0^L\psi(x)\sin\Bigl(\frac{n\pi x}{L}\Bigr)\,dx.
$$

---


## Quick Reference Summary

| Equation                        | Domain          | Method                          | Key formula |
|--------------------------------|-----------------|---------------------------------|-------------|
| Transport $u_t+cu_x=0$         | infinite line   | characteristics                 | $\phi(x-ct)$ |
| Transport + decay              | infinite line   | characteristics + ODE           | $\phi(x-ct)e^{-at}$ |
| Heat (pure IVP)                | infinite line   | Fourier transform / kernel      | convolution with Gaussian |
| Heat (Dirichlet)               | finite interval | separation of variables         | sine series with exponential decay |
| Wave (pure IVP)                | infinite line   | characteristics / d’Alembert    | average of left- and right-going waves |
| Wave (Dirichlet)               | finite interval | separation of variables         | sine series with cos/sin time factors |

All of the finite-domain formulas above rely on the same eigenvalue problem $X''+\lambda X=0$ with the given boundary conditions; only the time ODE changes from equation to equation.

# Concrete Worked Examples


## From the Transport Equation to a Dirichlet Problem  


*(with explicit functions and full calculations)*

We solve four related problems with the same style of initial data so you can see the methods in action with numbers and closed-form answers.

---



## Problem A – Transport Equation



Solve
$$
u_t + 2 u_x = 0, \qquad -\infty < x < \infty,\quad t > 0,
$$
$$
u(x,0) = e^{-x^2}.
$$

**Step-by-step**

1. Characteristics: $\frac{dx}{dt}=2$ $\Rightarrow$ $x-2t=\text{constant}$.

2. General solution: $u(x,t)=f(x-2t)$.

3. Initial condition: $u(x,0)=f(x)=e^{-x^2}$.

4. Therefore
$$
u(x,t)=e^{-(x-2t)^2}.
$$

**Check.**  
$u_t=-2(x-2t)(-2)e^{-(x-2t)^2}=4(x-2t)e^{-(x-2t)^2}$,  
$u_x=-2(x-2t)e^{-(x-2t)^2}$,  
so $u_t+2u_x=0$. ✓  
At $t=0$ we recover $e^{-x^2}$. ✓

---



## Problem B – Transport with Decay



Solve
$$
u_t + 2 u_x + u = 0, \qquad -\infty < x < \infty,\quad t > 0,
$$
$$
u(x,0)=e^{-x^2}.
$$

**Step-by-step**

1. Characteristics still $x-2t=\xi$.

2. Along a characteristic the PDE becomes the ODE $\frac{dv}{dt}+v=0$, so $v(t)=v(0)e^{-t}$.

3. Thus
$$
u(x,t)=e^{-(x-2t)^2}\,e^{-t}.
$$

**Check.**  
The extra factor $e^{-t}$ produces the $+u$ term that cancels the decay contribution. ✓

---



## Problem C – Heat Equation on the Infinite Line


Solve
$$
u_t = u_{xx}, \qquad -\infty < x < \infty,\quad t > 0,
$$
$$
u(x,0)=e^{-x^2}.
$$

**Step-by-step**

1. The fundamental solution (heat kernel) for $k=1$ is
$$
G(x,t)=\frac{1}{\sqrt{4\pi t}}\,e^{-x^2/(4t)}.
$$

2. The solution is the convolution
$$
u(x,t)=\int_{y \ = \ -\infty}^{\infty}G(x-y,t)\,e^{-y^2}\,dy.
$$

3. Completing the square in the exponent yields the exact closed form
$$
u(x,t)=\frac{1}{\sqrt{1+4t}}\,\exp\Bigl(-\frac{x^2}{1+4t}\Bigr).
$$

**Check.**  
At $t=0$ the prefactor is 1 and the exponential is $e^{-x^2}$. ✓  
Direct differentiation confirms it satisfies the heat equation.

---



## Problem D – Wave Equation on a Finite Interval (Dirichlet)



Solve
$$
u_{tt}=4u_{xx}, \qquad 0<x<\pi,\quad t>0,
$$
$$
u(0,t)=u(\pi,t)=0,
$$
$$
u(x,0)=\sin(2x), \qquad u_t(x,0)=0.
$$

**Step-by-step**

1. Separate variables: $u=X(x)T(t)$  
$$
\frac{T''}{4T}=\frac{X''}{X}=-\lambda.
$$

2. Boundary conditions $X(0)=X(\pi)=0$ give the eigenvalue problem  
$$
X''+\lambda X=0,\quad X(0)=X(\pi)=0.
$$
Solutions:
$$
\lambda_n=n^2,\qquad X_n(x)=\sin(nx),\quad n=1,2,3,\dots
$$

3. Time equation for each $n$:
$$
T''+4n^2 T=0\qquad\Rightarrow\qquad T_n(t)=A_n\cos(2nt)+B_n\sin(2nt).
$$

4. Superposition:
$$
u(x,t)=\sum_{n=1}^\infty\sin(nx)\bigl(A_n\cos(2nt)+B_n\sin(2nt)\bigr).
$$

5. Initial velocity $u_t(x,0)=0$ forces every $B_n=0$.

6. Initial displacement:
$$
\sum_{n=1}^\infty A_n\sin(nx)=\sin(2x).
$$
By orthogonality of the sines, $A_2=1$ and all other $A_n=0$.

7. Final answer:
$$
u(x,t)=\sin(2x)\cos(4t).
$$

**Check.**  
$u_{tt}=-16\sin(2x)\cos(4t)$,  
$4u_{xx}=4(-4)\sin(2x)\cos(4t)=-16\sin(2x)\cos(4t)$. ✓  
Boundary conditions: $\sin(0)=\sin(2\pi)=0$. ✓  
$u(x,0)=\sin(2x)$, $u_t(x,0)=-4\sin(2x)\sin(0)=0$. ✓

---



## Summary of the Concrete Answers



| Problem | Equation              | Initial data     | Explicit solution                          |
|---------|-----------------------|------------------|--------------------------------------------|
| A       | $u_t+2u_x=0$         | $e^{-x^2}$      | $e^{-(x-2t)^2}$                           |
| B       | $u_t+2u_x+u=0$       | $e^{-x^2}$      | $e^{-(x-2t)^2}e^{-t}$                     |
| C       | $u_t=u_{xx}$         | $e^{-x^2}$      | $\frac{1}{\sqrt{1+4t}}\exp\bigl(-x^2/(1+4t)\bigr)$ |
| D       | $u_{tt}=4u_{xx}$ (Dirichlet) | $\sin 2x$, $u_t=0$ | $\sin(2x)\cos(4t)$                    |

These four problems use the same initial profile (or a simple sine) so the calculations stay elementary while still illustrating every method from characteristics through separation of variables.
