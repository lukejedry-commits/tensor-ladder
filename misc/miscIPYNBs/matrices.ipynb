Note: This analysis is in 2-Dimensional spaces (as in, x and y; an orthonormal bases space)
$$$$
i.e. $\color{#5588ff} \bold{e}_1 = \hat{x} = \begin{bmatrix}1 \\ 0\end{bmatrix}$ and $\color{#d20000} \bold{e}_2 = \hat{y} = \begin{bmatrix}0 \\ 1\end{bmatrix}$
$$
\mathbf{A} = 
\begin{bmatrix}
A_{11} & A_{12} \\
A_{21} & A_{22}
\end{bmatrix}
$$
The matrix $\mathbf{A}$ has a characteristic property such that, given a generic vector $\vec{\mathbf{v}} = x \bold{e}_{1} + y \bold{e}_2$
$$\mathbf{A} \vec{\mathbf{v}} = \lambda \vec{\mathbf{v}}$$
Where $\lambda$ is the conventional Greek letter to describe the characteristic value(s), a.k.a. **eigenvalues** of the target matrix $\mathbf{A}$
$$$$
Though I'm restricting my analysis to 2-by-2 matrices, if one did analyze a generic __square__ $n$-by-$n$ matrix, then:
$$\operatorname{size}{\{\lambda\}} = n$$
That is to say, the number of $\lambda$ eigenvalues for said matrix would equal the number of column OR row entries, $n$.
$$$$
i.e. a 3-by-3 matrix has 3 eigenvalues $\left(\lambda_1, \lambda_2, \lambda_3\right)$; a 10-by-10 matrix has 10 eigenvalues $\left(\lambda_1, \lambda_2, \dots, \lambda_{10}\right)$
$$\text{tr}\left(\mathbf{A}\right) = A_{11} + A_{22}$$
$$\det{\left(\mathbf{A}\right)} = A_{11} A_{22} - A_{12} A_{21}$$
Eigenvalues $:=$
$$\det{\left(\mathbf{A}-\lambda \mathbf{I}\right)} = 0$$
$$\lambda_{1,2} = \frac{}{2}$$
