$$
\left(x * h\right)(t) = \int_{\tau \ = \ -\infty}^{\infty}{x\left(t-\tau\right)h\left(\tau\right) \ d\tau}
$$

# Continuous Convolution of Two Scalable Pulses

Algebraic / symbolic notes. Plots are a supplement only, after the closed form.

Assumptions throughout: $w_1 \ge w_2 > 0$. If your widths are swapped, rename them.

---

## Part 1 — Setup, overlap length, and the solution-path template

### Definitions

$$
U(t)=\begin{cases}1,&t>0\\0,&\text{otherwise}\end{cases},
\qquad
\Pi\!\left(\frac{x-c}{w}\right)
=U\!\left(x-c+\tfrac{w}{2}\right)
-U\!\left(x-c-\tfrac{w}{2}\right).
$$

So $\Pi((x-c)/w)=1$ exactly on the open interval $(c-w/2,\ c+w/2)$, and $0$ elsewhere. (Endpoints do not matter for the integral.)

Let

$$
f(x)=\Pi\!\left(\frac{x-c_1}{w_1}\right),\qquad
g(x)=\Pi\!\left(\frac{x-c_2}{w_2}\right),
$$

with the standing assumption

$$
w_1\ge w_2>0.
$$

### Convolution

$$
(f*g)(t)
=\int_{-\infty}^{\infty}f(\tau)\,g(t-\tau)\,d\tau.
$$

$f(\tau)=1$ only for $\tau\in I_f:=(a_1,b_1)$, where

$$
a_1=c_1-\tfrac{w_1}{2},\qquad b_1=c_1+\tfrac{w_1}{2}.
$$

$g(t-\tau)=1$ only when $c_2-w_2/2<t-\tau<c_2+w_2/2$. Solving for $\tau$:

$$
t-c_2-\tfrac{w_2}{2}<\tau<t-c_2+\tfrac{w_2}{2}.
$$

Call that moving interval $I_g(t):=(a_2(t),b_2(t))$,

$$
a_2(t)=t-c_2-\tfrac{w_2}{2},\qquad
b_2(t)=t-c_2+\tfrac{w_2}{2}.
$$

The integrand is $1$ exactly on $I_f\cap I_g(t)$, so

$$
(f*g)(t)
=\text{length}\bigl(I_f\cap I_g(t)\bigr)
=\max\bigl(0,\ \min(b_1,b_2(t))-\max(a_1,a_2(t))\bigr).
$$

That one formula is the whole problem; everything below is evaluating it by cases.

### Reusable solution-path template

(Any two compact-support indicator functions.)

1. Write each pulse as an interval $(a,b)$.
2. Write the second factor as a $t$-dependent interval.
3. Form $\max(0,\min(b\text{'s})-\max(a\text{'s}))$.
4. Find the finitely many $t$ where an $a$ of one meets a $b$ of the other (edge crossings). Those split the real line into regions.
5. In each region, drop the $\max/\min$ to a single expression and simplify.
6. Only then sketch: breakpoints and plateau height should match the algebra.

### Shift that kills the centers

Set

$$
u:=t-(c_1+c_2).
$$

Then after translating both pulses so their “sum of centers” is at the origin of $u$, one has $a_1=-w_1/2$, $b_1=w_1/2$. Equivalently: first solve the centered problem

$$
f_0=\Pi(x/w_1),\quad g_0=\Pi(x/w_2),
$$

get $(f_0*g_0)(u)$, then

$$
(f*g)(t)=(f_0*g_0)\bigl(t-(c_1+c_2)\bigr).
$$

The next part does the centered case exhaustively; centers return as that one substitution.

---

## Part 2 — Centered case $c_1=c_2=0$, $w_1\ge w_2>0$

Now $a_1=-w_1/2$, $b_1=w_1/2$, $a_2(t)=t-w_2/2$, $b_2(t)=t+w_2/2$.

### Step A — edge-crossing values of $t$

An edge of $I_g(t)$ meets an edge of $I_f$ when one of these holds:

$$
\begin{aligned}
t-\tfrac{w_2}{2}&=-\tfrac{w_1}{2}
&&\Rightarrow&& t=-\tfrac{w_1-w_2}{2},\\
t-\tfrac{w_2}{2}&=\tfrac{w_1}{2}
&&\Rightarrow&& t=\tfrac{w_1+w_2}{2},\\
t+\tfrac{w_2}{2}&=-\tfrac{w_1}{2}
&&\Rightarrow&& t=-\tfrac{w_1+w_2}{2},\\
t+\tfrac{w_2}{2}&=\tfrac{w_1}{2}
&&\Rightarrow&& t=\tfrac{w_1-w_2}{2}.
\end{aligned}
$$

Because $w_1\ge w_2$, these four numbers are ordered

$$
-\tfrac{w_1+w_2}{2}
\;\le\;
-\tfrac{w_1-w_2}{2}
\;\le\;
\tfrac{w_1-w_2}{2}
\;\le\;
\tfrac{w_1+w_2}{2}.
$$

(Equality of the middle two holds if and only if $w_1=w_2$; that collapse is treated separately later.)

That splits $\mathbb{R}$ into five regions. In each one, drop $\max/\min$ carefully.

---

### Region 0: $t\le -\dfrac{w_1+w_2}{2}$

Here $b_2(t)=t+w_2/2\le -w_1/2=a_1$, so $I_g$ lies entirely left of $I_f$. Intersection empty:

$$
(f*g)(t)=0.
$$

---

### Region 1 (rising ramp): $-\dfrac{w_1+w_2}{2}<t<-\dfrac{w_1-w_2}{2}$

Check the ordering of the four endpoints:

- $a_2(t)=t-w_2/2$: at the left endpoint of this region, $a_2=-(w_1+w_2)/2-w_2/2=-(w_1+2w_2)/2<-w_1/2$. At the right endpoint, $a_2=-(w_1-w_2)/2-w_2/2=-w_1/2$. So throughout the open region, $a_2<a_1$.
- $b_2(t)=t+w_2/2$: at the left endpoint $b_2=-w_1/2=a_1$; it increases, and at the right endpoint $b_2=-(w_1-w_2)/2+w_2/2=-w_1/2+w_2$. Compare to $b_1=w_1/2$: $-w_1/2+w_2\le w_1/2$ iff $w_2\le w_1$, which holds. So $a_1<b_2\le b_1$ in this region (strict on the open interval).

Thus

$$
\max(a_1,a_2)=a_1,\qquad\min(b_1,b_2)=b_2,
$$

and

$$
(f*g)(t)=b_2-a_1=t+\tfrac{w_2}{2}-\Bigl(-\tfrac{w_1}{2}\Bigr)
=t+\tfrac{w_1+w_2}{2}.
$$

Sanity: at left end $\to 0$; at right end $\to w_2$.

---

### Region 2 (flat top): $-\dfrac{w_1-w_2}{2}\le t\le\dfrac{w_1-w_2}{2}$

Now $a_2\ge a_1$ (because $t-w_2/2\ge -w_1/2$ iff $t\ge -(w_1-w_2)/2$), and $b_2\le b_1$ (same check flipped). Also $a_2<b_2$ always. So the whole of $I_g$ sits inside $I_f$:

$$
\max(a_1,a_2)=a_2,\qquad\min(b_1,b_2)=b_2,
$$

$$
(f*g)(t)=b_2-a_2=w_2.
$$

Plateau height is exactly the shorter width. (This interval has positive length only when $w_1>w_2$.)

---

### Region 3 (falling ramp): $\dfrac{w_1-w_2}{2}<t<\dfrac{w_1+w_2}{2}$

Symmetric to Region 1:

- $b_2>b_1$ throughout the open region,
- $a_2$ still satisfies $a_1\le a_2<b_1$.

So

$$
\max(a_1,a_2)=a_2,\qquad\min(b_1,b_2)=b_1,
$$

$$
(f*g)(t)=b_1-a_2
=\tfrac{w_1}{2}-\Bigl(t-\tfrac{w_2}{2}\Bigr)
=\tfrac{w_1+w_2}{2}-t.
$$

Sanity: at left end $\to w_2$; at right end $\to 0$.

---

### Region 4: $t\ge\dfrac{w_1+w_2}{2}$

Here $a_2\ge b_1$, no overlap:

$$
(f*g)(t)=0.
$$

---

## Centered closed form ($w_1 > w_2$)

$$
(f_0*g_0)(t)=
\begin{cases}
0,& |t|\ge\dfrac{w_1+w_2}{2},\\[6pt]
t+\dfrac{w_1+w_2}{2},&
-\dfrac{w_1+w_2}{2}<t<-\dfrac{w_1-w_2}{2},\\[6pt]
w_2,&
|t|\le\dfrac{w_1-w_2}{2},\\[6pt]
\dfrac{w_1+w_2}{2}-t,&
\dfrac{w_1-w_2}{2}<t<\dfrac{w_1+w_2}{2}.
\end{cases}
$$

Compact description: a trapezoid of height $w_2$, bottom base $w_1+w_2$, top base $w_1-w_2$.

---

# Continuous Convolution of Two Scalable Pulses

Algebraic / symbolic notes. Allowed tools: the unit step $U$, the pulse $\Pi$, and the definite integral. No $\min$ / $\max$. Plots are a supplement only, after the closed form.

Assumptions throughout: $w_1 \ge w_2 > 0$. If your widths are swapped, rename them.

---

## Part 1 — Setup (only $U$, $\Pi$, and the integral)

### Definitions

$$
U(t)=\begin{cases}1,&t>0\\0,&\text{otherwise}\end{cases},
\qquad
\Pi\!\left(\frac{x-c}{w}\right)
=U\!\left(x-c+\tfrac{w}{2}\right)
-U\!\left(x-c-\tfrac{w}{2}\right).
$$

So $\Pi((x-c)/w)=1$ exactly on the open interval $(c-w/2,\ c+w/2)$, and $0$ elsewhere. (Endpoints do not matter for the integral.)

Let

$$
f(x)=\Pi\!\left(\frac{x-c_1}{w_1}\right),\qquad
g(x)=\Pi\!\left(\frac{x-c_2}{w_2}\right),
$$

with

$$
w_1\ge w_2>0.
$$

### Convolution as an integral of a product of pulses

$$
(f*g)(t)
=\int_{-\infty}^{\infty}f(\tau)\,g(t-\tau)\,d\tau
=\int_{-\infty}^{\infty}
\Pi\!\left(\frac{\tau-c_1}{w_1}\right)
\Pi\!\left(\frac{t-\tau-c_2}{w_2}\right)
\,d\tau.
$$

The first factor is $1$ only for

$$
c_1-\tfrac{w_1}{2}<\tau<c_1+\tfrac{w_1}{2}.
$$

The second factor is $1$ only when

$$
c_2-\tfrac{w_2}{2}<t-\tau<c_2+\tfrac{w_2}{2}.
$$

Solving the second inequality for $\tau$ (subtract $t$, then multiply by $-1$ and reverse both inequalities):

$$
t-c_2-\tfrac{w_2}{2}<\tau<t-c_2+\tfrac{w_2}{2}.
$$

So the product is $1$ only where $\tau$ lies in **both** open intervals at once, and is $0$ elsewhere. Therefore the infinite integral collapses to a finite integral of $1$:

$$
(f*g)(t)
=\int_{L(t)}^{R(t)} 1\,d\tau
=R(t)-L(t),
$$

**provided** the two intervals overlap (i.e. the lower limit is strictly less than the upper). If they do not overlap, the product is identically zero and $(f*g)(t)=0$.

Explicitly, the lower and upper limits of the surviving integral are the right-hand left-edge and the left-hand right-edge of the two intervals:

- left edge of $f$: $A_1:=c_1-w_1/2$
- right edge of $f$: $B_1:=c_1+w_1/2$
- left edge of the $g$-slice: $A_2(t):=t-c_2-w_2/2$
- right edge of the $g$-slice: $B_2(t):=t-c_2+w_2/2$

On a region where we already know which edge is innermost, we will write $L(t)$ and $R(t)$ as one of $\{A_1,A_2(t)\}$ and one of $\{B_1,B_2(t)\}$ respectively, then integrate. We never need a $\min$/$\max$ symbol for that: the region itself tells us which edge is which.

### Shift that kills the centers

Set

$$
u:=t-(c_1+c_2).
$$

Equivalently: first solve the centered problem

$$
f_0=\Pi\!\left(\frac{u}{w_1}\right),\qquad
g_0=\Pi\!\left(\frac{u}{w_2}\right),
$$

get $(f_0*g_0)(u)$, then

$$
(f*g)(t)=(f_0*g_0)\bigl(t-(c_1+c_2)\bigr).
$$

Part 2 does the centered case. Centers return as that one substitution (Part 3, later).

### Reusable solution-path template (no $\min$/$\max$)

1. Write $(f*g)(t)$ as $\displaystyle\int\Pi((\tau-c_1)/w_1)\,\Pi((t-\tau-c_2)/w_2)\,d\tau$.
2. From each $\Pi$, read off the open interval where that factor equals $1$.
3. Find every $t$ at which an edge of the $f$-interval meets an edge of the $g$-slice (four crossings). Those values split $\mathbb{R}$ into regions.
4. In each region, decide by direct comparison which endpoint is the lower limit $L(t)$ and which is the upper limit $R(t)$.
5. If $L(t)<R(t)$, evaluate $\displaystyle\int_{L(t)}^{R(t)}1\,d\tau=R(t)-L(t)$. If not, the integral is $0$.
6. Only then sketch: breakpoints and plateau height should match the algebra.

---

## Part 2 — Centered case $c_1=c_2=0$, $w_1\ge w_2>0$

Now

$$
(f*g)(t)
=\int_{-\infty}^{\infty}
\Pi\!\left(\frac{\tau}{w_1}\right)
\Pi\!\left(\frac{t-\tau}{w_2}\right)
\,d\tau,
$$

and the two intervals are

$$
-\tfrac{w_1}{2}<\tau<\tfrac{w_1}{2},
\qquad
t-\tfrac{w_2}{2}<\tau<t+\tfrac{w_2}{2}.
$$

### Step A — edge-crossing values of $t$

Set each edge of the second interval equal to each edge of the first:

$$
\begin{aligned}
t-\tfrac{w_2}{2}&=-\tfrac{w_1}{2}
&&\Rightarrow&& t=-\tfrac{w_1-w_2}{2},\\
t-\tfrac{w_2}{2}&=\tfrac{w_1}{2}
&&\Rightarrow&& t=\tfrac{w_1+w_2}{2},\\
t+\tfrac{w_2}{2}&=-\tfrac{w_1}{2}
&&\Rightarrow&& t=-\tfrac{w_1+w_2}{2},\\
t+\tfrac{w_2}{2}&=\tfrac{w_1}{2}
&&\Rightarrow&& t=\tfrac{w_1-w_2}{2}.
\end{aligned}
$$

Because $w_1\ge w_2$, these four numbers are ordered

$$
-\tfrac{w_1+w_2}{2}
\;\le\;
-\tfrac{w_1-w_2}{2}
\;\le\;
\tfrac{w_1-w_2}{2}
\;\le\;
\tfrac{w_1+w_2}{2}.
$$

(Equality of the middle two holds if and only if $w_1=w_2$; that collapse is treated separately later.)

That splits $\mathbb{R}$ into five regions. In each region we write a separate integral.

---

### Region 0: $t\le -\dfrac{w_1+w_2}{2}$

Compare the right edge of the $g$-slice to the left edge of $f$:

$$
t+\tfrac{w_2}{2}\le -\tfrac{w_1}{2}.
$$

So every $\tau$ with $\tau<t+w_2/2$ also satisfies $\tau\le -w_1/2$, which is outside the support of $\Pi(\tau/w_1)$. The product is identically $0$:

$$
(f*g)(t)
=\int_{-\infty}^{\infty}
\Pi\!\left(\frac{\tau}{w_1}\right)
\Pi\!\left(\frac{t-\tau}{w_2}\right)
\,d\tau
=0.
$$

---

### Region 1 (rising ramp): $-\dfrac{w_1+w_2}{2}<t<-\dfrac{w_1-w_2}{2}$

**Which edges bound the overlap?**

- Left edge of the $g$-slice vs left edge of $f$:
  $$
  t-\tfrac{w_2}{2}
  < -\tfrac{w_1-w_2}{2}-\tfrac{w_2}{2}
  = -\tfrac{w_1}{2},
  $$
  so $t-w_2/2 < -w_1/2$. The more restrictive (larger) lower bound is therefore $-w_1/2$.
- Right edge of the $g$-slice vs right edge of $f$:
  $$
  t+\tfrac{w_2}{2}
  < -\tfrac{w_1-w_2}{2}+\tfrac{w_2}{2}
  = -\tfrac{w_1}{2}+w_2.
  $$
  And $-w_1/2+w_2\le w_1/2$ because $w_2\le w_1$. Also $t+w_2/2 > -w_1/2$ throughout the open region (that is exactly the left boundary of Region 1). So the more restrictive (smaller) upper bound is $t+w_2/2$.

Thus the product equals $1$ precisely for

$$
-\tfrac{w_1}{2}<\tau<t+\tfrac{w_2}{2},
$$

and the convolution reduces to

$$
(f*g)(t)
=\int_{-w_1/2}^{\,t+w_2/2} 1\,d\tau
=\Bigl[\tau\Bigr]_{-w_1/2}^{t+w_2/2}
=\Bigl(t+\tfrac{w_2}{2}\Bigr)-\Bigl(-\tfrac{w_1}{2}\Bigr)
=t+\tfrac{w_1+w_2}{2}.
$$

Sanity: as $t\to -(w_1+w_2)/2$ from above, the value $\to 0$; as $t\to -(w_1-w_2)/2$ from below, the value $\to w_2$.

---

### Region 2 (flat top): $-\dfrac{w_1-w_2}{2}\le t\le\dfrac{w_1-w_2}{2}$

**Which edges bound the overlap?**

- Lower bounds: $t-w_2/2 \ge -w_1/2$ (definition of the left end of this region), so the more restrictive lower bound is $t-w_2/2$.
- Upper bounds: $t+w_2/2 \le w_1/2$ (definition of the right end), so the more restrictive upper bound is $t+w_2/2$.

Thus the entire $g$-slice sits inside the $f$-interval, and

$$
(f*g)(t)
=\int_{t-w_2/2}^{\,t+w_2/2} 1\,d\tau
=\Bigl[\tau\Bigr]_{t-w_2/2}^{t+w_2/2}
=\bigl(t+\tfrac{w_2}{2}\bigr)-\bigl(t-\tfrac{w_2}{2}\bigr)
=w_2.
$$

Plateau height is exactly the shorter width. (This region has positive length only when $w_1>w_2$.)

---

### Region 3 (falling ramp): $\dfrac{w_1-w_2}{2}<t<\dfrac{w_1+w_2}{2}$

**Which edges bound the overlap?**

- Lower bounds: $t-w_2/2 > (w_1-w_2)/2 - w_2/2 = w_1/2 - w_2$. And $w_1/2-w_2 \ge -w_1/2$ because $w_2\le w_1$. Also $t-w_2/2 < w_1/2$ throughout the open region. So the more restrictive lower bound is $t-w_2/2$.
- Upper bounds: $t+w_2/2 > (w_1-w_2)/2 + w_2/2 = w_1/2$, so $t+w_2/2 > w_1/2$. The more restrictive upper bound is $w_1/2$.

Thus

$$
(f*g)(t)
=\int_{t-w_2/2}^{\,w_1/2} 1\,d\tau
=\Bigl[\tau\Bigr]_{t-w_2/2}^{w_1/2}
=\tfrac{w_1}{2}-\Bigl(t-\tfrac{w_2}{2}\Bigr)
=\tfrac{w_1+w_2}{2}-t.
$$

Sanity: as $t\to (w_1-w_2)/2$ from above, the value $\to w_2$; as $t\to (w_1+w_2)/2$ from below, the value $\to 0$.

---

### Region 4: $t\ge\dfrac{w_1+w_2}{2}$

Compare the left edge of the $g$-slice to the right edge of $f$:

$$
t-\tfrac{w_2}{2}\ge \tfrac{w_1}{2}.
$$

So every $\tau$ with $\tau>t-w_2/2$ also satisfies $\tau\ge w_1/2$, outside the support of $\Pi(\tau/w_1)$. The product is identically $0$:

$$
(f*g)(t)=0.
$$

---

## Summary: the five integrals (centered, $w_1>w_2$)

$$
(f_0*g_0)(t)=
\begin{cases}
\displaystyle\int_{-\infty}^{\infty}
\Pi\!\left(\frac{\tau}{w_1}\right)
\Pi\!\left(\frac{t-\tau}{w_2}\right)\,d\tau
=0,
& t\le -\dfrac{w_1+w_2}{2},
\\[14pt]
\displaystyle\int_{-w_1/2}^{\,t+w_2/2} 1\,d\tau
=t+\dfrac{w_1+w_2}{2},
& -\dfrac{w_1+w_2}{2}<t<-\dfrac{w_1-w_2}{2},
\\[14pt]
\displaystyle\int_{t-w_2/2}^{\,t+w_2/2} 1\,d\tau
=w_2,
& |t|\le\dfrac{w_1-w_2}{2},
\\[14pt]
\displaystyle\int_{t-w_2/2}^{\,w_1/2} 1\,d\tau
=\dfrac{w_1+w_2}{2}-t,
& \dfrac{w_1-w_2}{2}<t<\dfrac{w_1+w_2}{2},
\\[14pt]
0,
& t\ge\dfrac{w_1+w_2}{2}.
\end{cases}
$$

Closed form (same thing, without writing the integrals):

$$
(f_0*g_0)(t)=
\begin{cases}
0,& |t|\ge\dfrac{w_1+w_2}{2},\\[6pt]
t+\dfrac{w_1+w_2}{2},&
-\dfrac{w_1+w_2}{2}<t<-\dfrac{w_1-w_2}{2},\\[6pt]
w_2,&
|t|\le\dfrac{w_1-w_2}{2},\\[6pt]
\dfrac{w_1+w_2}{2}-t,&
\dfrac{w_1-w_2}{2}<t<\dfrac{w_1+w_2}{2}.
\end{cases}
$$

Compact description: a trapezoid of height $w_2$, bottom base $w_1+w_2$, top base $w_1-w_2$.
