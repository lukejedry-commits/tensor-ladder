### Given:

$$
u_{tt} = c^2 u_{xx} \ \text{where }
\begin{cases}
u(t,0) &= \phi(t) \\
u(t,L) &= \psi(t)
\end{cases}
$$
$$\forall \ x: \{x \in \R,0\le x\le L\}, \quad t: \{t \in \R, t\ge 0\}$$

### Find:

$$u(t,x) = \ ?$$

### Assume:

 - When two functions of separate independent variables equal each other (entirely absent of the other independent variable on each side of the `=` sign), set said functions equal to $-\lambda$
 - $u(t,x)$ is separable; $$u(t,x) = T(t)\,X(x)$$
 - $$u_{q_{i}} = \frac{\partial u}{\partial q_{i}} \quad u_{q_i q_j} = \frac{\partial^2 u}{\partial q_j q_i}$$

### Solution:

#### Part One: Separate

$$u_{tt} = c^2 u_{xx}$$
$$\frac{\partial}{\partial t^2}\left[T(t)X(x)\right] = c^2 \frac{\partial}{\partial x^2}\left[T(t)X(x)\right]$$
$$\frac{\partial}{\partial t}\left[\frac{\partial}{\partial t}\left[T(t)X(x)\right]\right] = c^2 \frac{\partial}{\partial x}\left[\frac{\partial}{\partial x}\left[T(t)X(x)\right]\right]$$
$$\frac{\partial}{\partial t}\left[T'(t)X(x)+ \cancel{0\cdot T(t)}\right] = c^2 \frac{\partial}{\partial x}\left[T(t)X'(x) + \cancel{0\cdot X(x)}\right]$$
$$\frac{\partial}{\partial t}\left[T'(t)X(x)\right] = c^2 \frac{\partial}{\partial x}\left[T(t)X'(x)\right]$$
$$T''(t)X(x) + \cancel{0\cdot T'(t)} = c^2 \left[T(t)X''(x) + \cancel{0\cdot X'(x)}\right]$$
$$T''(t)X(x) = c^2 T(t)X''(x)$$
$$\frac{1}{c^2 T(t)\cancel{X(x)}}\left[T''(t)\cancel{X(x)}\right] = \left[\cancel{c^2 T(t)}X''(x)\right]\frac{1}{\cancel{c^2 T(t)}X(x)}$$
$$\frac{T''}{c^2 T}=\frac{X''}{X}$$
$$\therefore \frac{T''}{c^2 T}=\frac{X''}{X}=-\lambda$$

#### Part Two: Isolate Ind. Var. 1

More is known for $X$, solve what you can with it first.
$$X\left(\frac{X''}{X}\right) = \left(-\lambda\right)X$$
$$\lambda X + X'' = \cancel{-\lambda X + \lambda X}$$
$$X'' + \lambda X = 0$$

Guess $X(x) = A e^{ax}$
$$\because X(x) = A e^{ax} \therefore X''(x) = A a^2 e^{ax}$$
$$\therefore A a^2 e^{ax} + A \lambda e^{ax} = 0$$
$$\therefore A e^{ax} \left(a^2 + \lambda \right) = 0$$
$$A = 0 \ \text{is a trivial solution.}$$
$$e^{ax} = 0 \ \text{is an impossible solution for any values of } a \text{ and } x \text{ or their combinations, }ax$$
$$\therefore a^2 + \lambda = 0 \Rightarrow a = \pm i \sqrt{\lambda} \Rightarrow a_{1,2} = +i\sqrt{\lambda}, \ -i\sqrt{\lambda}$$

$$\therefore X(x) = A_1 e^{i \sqrt{\lambda} x} + A_2 e^{-i \sqrt{\lambda} x}$$
$$X(0) = \phi(t) = A_1 e^{0} + A_2 e^{0} \therefore A_1 + A_2 = \phi$$
$$X(L) = \psi(t) = i L \sqrt{\lambda} A_1 e^{i L\sqrt{\lambda}} - i L \sqrt{\lambda} A_2 e^{-i L\sqrt{\lambda}}$$
$$\therefore i L \sqrt{\lambda}\left(A_1 e^{i L\sqrt{\lambda}} - A_2 e^{-i L\sqrt{\lambda}}\right) = \psi$$
$$\Rightarrow A_1 = \phi - A_2$$
$$\Rightarrow iL\sqrt{\lambda}\left(\left(\phi - A_2\right) e^{i L\sqrt{\lambda}} - A_2 e^{-i L\sqrt{\lambda}}\right) = \psi \quad$$
$$\Rightarrow \psi = iL\sqrt{\lambda}\left(\phi e^{iL\sqrt{\lambda}} - A_2 \left(e^{i L\sqrt{\lambda}}+e^{-i L\sqrt{\lambda}}\right)\right)$$

$$\psi = iL\sqrt{\lambda}\left(\phi e^{iL\sqrt{\lambda}} - 2A_2 \left(\frac{e^{iL\sqrt{\lambda}}+e^{-iL\sqrt{\lambda}}}{2}\right)\right)$$
$$\phi e^{i L\sqrt{\lambda}} - 2A_2 \cos\left(L\sqrt{\lambda}\right) = -i \frac{1}{L\sqrt{\lambda}}\psi$$
$$2A_2 \cos\left(L\sqrt{\lambda}\right) = \phi e^{iL\sqrt{\lambda}} + i \frac{1}{L\sqrt{\lambda}}\psi$$


$$A_2 = \frac{1}{2\sqrt{\lambda}}\sec\left(L\sqrt{\lambda}\right)\left(\sqrt{\lambda}\phi e^{iL\sqrt{\lambda}} + i \psi \right)$$

$$A_1 = \phi - \frac{1}{2\sqrt{\lambda}}\sec\left(L\sqrt{\lambda}\right)\left(\sqrt{\lambda} \phi e^{iL\sqrt{\lambda}} + i\psi\right)$$

$$A_1 = \frac{1}{2\sqrt{\lambda}}\sec\left(L\sqrt{\lambda}\right)\left[2\sqrt{\lambda}\phi\cos\left(L\sqrt{\lambda}\right)+\sqrt{\lambda}\phi e^{iL\sqrt{\lambda}} + i\psi\right]$$
$$A_1 = \frac{1}{2\lambda}\sec\left(L\sqrt{\lambda}\right)\left[\lambda\phi\left(2\cos\left(L\sqrt{\lambda}\right) + e^{i L\sqrt{\lambda}}\right)+i\psi\sqrt{\lambda}\right]$$
