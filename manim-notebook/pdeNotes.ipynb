### Given:

$$
u_{tt} = c^2 u_{xx} \ \text{where }
\begin{cases}
u(t,0) &= \phi(t) \\
u(t,L) &= \psi(t)
\end{cases}
$$
$$\forall \ x: \{x \in \R,0\le x\le L\}, \quad t: \{t \in \R, t\ge 0\}$$

### Find:

$$u(t,x) = \ ?$$

### Assume:

 - When two functions of separate independent variables equal each other (entirely absent of the other independent variable on each side of the `=` sign), set said functions equal to $-\lambda$
 - $u(t,x)$ is separable; $$u(t,x) = T(t)\,X(x)$$
 - $$u_{q_{i}} = \frac{\partial u}{\partial q_{i}} \quad u_{q_i q_j} = \frac{\partial^2 u}{\partial q_j q_i}$$

### Solution:

#### Part One: Separate

$$u_{tt} = c^2 u_{xx}$$
$$\frac{\partial}{\partial t^2}\left[T(t)X(x)\right] = c^2 \frac{\partial}{\partial x^2}\left[T(t)X(x)\right]$$
$$\frac{\partial}{\partial t}\left[\frac{\partial}{\partial t}\left[T(t)X(x)\right]\right] = c^2 \frac{\partial}{\partial x}\left[\frac{\partial}{\partial x}\left[T(t)X(x)\right]\right]$$
$$\frac{\partial}{\partial t}\left[T'(t)X(x)+ \cancel{0\cdot T(t)}\right] = c^2 \frac{\partial}{\partial x}\left[T(t)X'(x) + \cancel{0\cdot X(x)}\right]$$
$$\frac{\partial}{\partial t}\left[T'(t)X(x)\right] = c^2 \frac{\partial}{\partial x}\left[T(t)X'(x)\right]$$
$$T''(t)X(x) + \cancel{0\cdot T'(t)} = c^2 \left[T(t)X''(x) + \cancel{0\cdot X'(x)}\right]$$
$$T''(t)X(x) = c^2 T(t)X''(x)$$
$$\frac{1}{c^2 T(t)\cancel{X(x)}}\left[T''(t)\cancel{X(x)}\right] = \left[\cancel{c^2 T(t)}X''(x)\right]\frac{1}{\cancel{c^2 T(t)}X(x)}$$
$$\frac{T''}{c^2 T}=\frac{X''}{X}$$
$$\therefore \frac{T''}{c^2 T}=\frac{X''}{X}=-\lambda$$

#### Part Two: Isolate Ind. Var. 1

More is known for $X$, solve what you can with it first.
$$X\left(\frac{X''}{X}\right) = \left(-\lambda\right)X$$
$$\lambda X + X'' = \cancel{-\lambda X + \lambda X}$$
$$X'' + \lambda X = 0$$

Guess $X(x) = A e^{ax}$
$$\because X(x) = A e^{ax} \therefore X''(x) = A a^2 e^{ax}$$
$$\therefore A a^2 e^{ax} + A \lambda e^{ax} = 0$$
$$\therefore A e^{ax} \left(a^2 + \lambda \right) = 0$$
$$A = 0 \ \text{is a trivial solution.}$$
$$e^{ax} = 0 \ \text{is an impossible solution for any values of } a \text{ and } x$$
$$\therefore a^2 + \lambda = 0 \Rightarrow a = \pm i \lambda \Rightarrow a_{1,2} = +i\lambda, \ -i\lambda$$

$$\therefore X(x) = A_1 e^{i \lambda x} + A_2 e^{-i \lambda x}$$
$$X(0) = \phi(t) = A_1 e^{0} + A_2 e^{0} \therefore A_1 + A_2 = \phi$$
$$X(L) = \psi(t) = i \lambda A_1 e^{i \lambda L} - i \lambda A_2 e^{-i \lambda L} \therefore i \lambda\left(A_1 e^{i \lambda L} - A_2 e^{-i \lambda L}\right) = \psi$$
$$\Rightarrow A_1 = \phi - A_2$$
$$\Rightarrow i \lambda\left(\left(\phi - A_2\right) e^{i \lambda L} - A_2 e^{-i \lambda L}\right) = \psi \quad \Rightarrow \psi = i\lambda\left(\phi e^{i\lambda L} - A_2 \left(e^{i\lambda L}+e^{-i\lambda L}\right)\right)$$

$$\psi = i\lambda\left(\phi e^{i\lambda L} - 2A_2 \left(\frac{e^{i\lambda L}+e^{-i\lambda L}}{2}\right)\right)$$
$$\phi e^{i\lambda L} - 2A_2 \cos\left(\lambda L\right) = -i \frac{1}{\lambda}\psi$$
$$2A_2 \cos\left(\lambda L\right) = \phi e^{i\lambda L} + i \frac{1}{\lambda}\psi$$


$$A_2 = \frac{1}{2\lambda}\sec\left(\lambda L\right)\left(\lambda\phi e^{i\lambda L} + i \psi \right)$$

$$A_1 = \phi - \frac{1}{2\lambda}\sec\left(\lambda L\right)\left(\lambda \phi e^{i\lambda L} + i\psi\right)$$

$$A_1 = \frac{1}{2\lambda}\sec\left(\lambda L\right)\left[2\lambda\phi\cos\left(\lambda L\right)+\lambda\phi e^{i\lambda L} + i\psi\right]$$
$$A_1 = \frac{1}{2\lambda}\sec\left(\lambda L\right)\left[\lambda\phi\left(2\cos\left(\lambda L\right) + e^{i \lambda L}\right)+i\psi\right]$$
